```python
#Convert Join_Date to datetime
df['Join_Date']=pd.to_datetime(df['Join_Date'],errors='coerce',format='%d-%m-%Y')
df
```

```python
#Split 'Department_Region' into 'Department' and 'Region'
df[['Department', 'Region']] = df['Department_Region'].str.split('-', expand=True)
df
```

```python
# Handle missing values (Impute with Median)
df['Age'] = df['Age'].fillna(df['Age'].median()).astype(int)
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
df
```

```python
# Check for duplicate rows
print("Duplicate rows count:", df.duplicated().sum())
df
```

```python
df.info()
df
```

```python
# Fix Phone column: convert to positive string numbers
df['Phone'] = df['Phone'].abs().astype(str)
df
```

```python
#  Drop the redundant combined column
df.drop(columns=['Department_Region'], inplace=True)
df
```

```python
# 3. Create a clean Full_Name column 
df['Full_Name'] = df['First_Name'] + ' ' + df['Last_Name']
df
```
